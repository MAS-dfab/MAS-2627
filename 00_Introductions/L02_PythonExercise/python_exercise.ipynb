## Go through the inventory!

## Exercise 1: Pascal's triangle
Pascal’s triangle is a geometric arrangement of numbers where each number is the sum of the two numbers directly above it.

<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/py_exercise/content/images/pascal_triangle.svg" width="320">

```python
#Practice Pascal triangle

def pascal_triangle(row):
    start = 1
    ori_pascal_row = []
    for i in range(row):
        nexten_pascal_row =[]
        if i == 0:
            nexten_pascal_row.append(start)
        else:
            nexten_pascal_row.append(start)
            for i in range(len(ori_pascal_row)-1):
                current = ori_pascal_row[i]
                next = ori_pascal_row[i+1]
                number = current + next 
                nexten_pascal_row.append(number)
            nexten_pascal_row.append(start)
        print(nexten_pascal_row)
        ori_pascal_row = nexten_pascal_row


pascal_triangle(10)
```

```python
def pairwise(iterable):
    # pairwise('ABCDEFG') → AB BC CD DE EF FG

    iterator = iter(iterable) #把「可迭代物件」轉成「迭代器(iterator)」
    a = next(iterator, None) #跟迭代器要下一個值

    for b in iterator: 
        yield a, b #generator(生成器) 的關鍵字
        a = b #下一輪迴圈
```

```python
from itertools import pairwise

row = [1,2,1]
next_row = [1]
for a, b in pairwise(row):
    next_row.append(a + b)

next_row.append(1)
print (next_row)
```

#### Create a function
New in town, **return**. Sends a value back to wherever the function was called.

```python
def next_pascal_row(row):
    next_row = [1]
    for a, b in pairwise(row):
        next_row.append(a + b)
    next_row.append(1)
    return next_row  #output

# list comprehension version of the next_pascal_row function
def next_pascal_row_comp(row):
    return [1] + [a + b for a, b in pairwise(row)] + [1]

num = 7
row = [1]
print (row)
for i in range(num):
    row = next_pascal_row(row)
    print (row)
```

#### Recursion(遞迴)
Recursion is when a function calls itself to solve a smaller version of the same problem. This continues until it reaches a simple case that can be solved directly, called the base case. The answers are then returned step by step to build the final result.
[Visualization](https://preview.redd.it/i-restored-in-hd-4k-the-original-spider-man-pointing-at-v0-bekphnqftcb41.jpg?width=1080&crop=smart&auto=webp&s=51243691bab29346bb8a94896cf6edba598de3c7)

```python
# recursive version of the next_pascal_row function
#elegant

def pascal(n):
    if n == 0:
        return [1]

    previous = pascal(n - 1) ####
    print (previous)

    return [1] + [a + b for a, b in pairwise(previous)] + [1]

pascal(4)
```

#### Try it youself
Use what we've just learned and create a [**Fibonacci**](https://en.wikipedia.org/wiki/Fibonacci_sequence) or [**Collatz conjecture**](https://en.wikipedia.org/wiki/Collatz_conjecture) generator!

```python
# Fibonacci

def fabonacci(n):
    if n ==0:
        return [0]
    
    if n ==1:
        return [0, 1]
    
    pre_list = fabonacci(n-1)
    previous = pre_list+[pre_list[-1]+pre_list[-2]]

    return previous
    
print(fabonacci(10))
```

```python
#Collatz Conjecture

def Collatz(n):

    if n == 1:
        return [1]
    
    if n %2 == 0:
        k = n//2
    else:
        k =  3*n+1

    list = [n]+ Collatz(k)

    return list

num = 20

message = f"Recursion ran {len(Collatz(num))} times"
print(Collatz(num), message)
```

```python
#While 

def C(n):
    next_list = [n]
    while n != 1:
        if n %2 == 0:
            n = n//2
        else:
            n =  3*n+1
        
        next_list.append(n)
        #while 迴圈一定要有「讓條件終將改變」的那一行

    return next_list


print(C(10))
```
