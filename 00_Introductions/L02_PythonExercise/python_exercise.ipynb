## A sequential collection of things
**list** -> [0,1,2,3]\
**tuple** -> (0,1,2)

```python
imlist = [".", ":", "-", "=", "+", "*", "#"]
print (type(imlist))

imtuple = (".", ":", "-", "=", "+", "*", "#")
print (type(imtuple))
```

#### What's the different between list and tuple?
Tuple is a **fixed size** and **immutable** collection of things

```python
# fixed size means we cannot add or remove elements from a tuple.
imlist.append("!")
imtuple.append("!") 
```

```python
# immutable means not changeable, so we cannot change the elements of a tuple.
imlist[0] = "!"
print (imlist)
imtuple[0] = "!"
```

## Again?
Loop, a repeated action. Like a *for* loop  

```python
for i in range(5):
    print (5-i)
print ("Happy New Year!")
```

Another kinds of loop is a *while* loop, which stop whenever condition is met

```python
count_down = 5
while count_down > 0:
    print (count_down)
    count_down -= 1 # try commenting this line to see what happens
print ("Happy New Year!")
```

## A slice of _?
To access part of the collection is called list slicing\
list [ **Start** : **End** : **Step** ]

```python
name = "Ananya"

# spell it out
print ("spelling out the name:")
for letter in name:
    print (letter)

# pronounce it
print ("pronouncing the name:")
print (name[0])
print (name[1:3])
print (name[3:])

# now spell it out in reverse
print ("spelling out the name in reverse:")
for letter in name[::-1]:
    print (letter)
```

## Go through the inventory!
**enumerate()** and **zip()**. Handy tools while using a for loop

```python
first_name = ['Ananya', 'Chewei']
last_name = ['Kango', 'Lin']

# enumerate() function is used to iterate over a list and get the index of each element.
print ("Enumerating the first names:")
for i, vor_name in enumerate(first_name):
    print (i, vor_name)

# zip() function is used to combine two lists into a single iterable object.
print ("Zipping the first and last names:")
for vor_name, nach_name in zip(first_name, last_name):
    print (vor_name, nach_name)

print ("Enumerating and zipping the first and last names:")
# combine enumerate() and zip() to get the index and the elements of two lists at the same time.
for i, (vor_name, nach_name) in enumerate(zip(first_name, last_name)):
    print (i, vor_name, nach_name)
```

## Exercise 1: Pascal's triangle
Pascal’s triangle is a geometric arrangement of numbers where each number is the sum of the two numbers directly above it.
<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/master/content/images/pascal_triangle.svg" width="320">

```python
# calculate the next row of Pascal's triangle given the current row
row = [1,2,1]
next_row = [1]

for i, num in enumerate(row[:-1]):
    next_row.append(num + row[i+1])

next_row.append(1)
print (next_row)
```

#### Advantage of python: Useful packages!
introducing ***itertools***

find out more about [itertools](https://docs.python.org/3/library/itertools.html)


```python
from itertools import pairwise

row = [1,2,1]
next_row = [1]
for a, b in pairwise(row):
    next_row.append(a + b)

next_row.append(1)
print (next_row)
```

#### Create a function
New in town, **return**. Sends a value back to wherever the function was called.

```python
def next_pascal_row(row):
    next_row = [1]
    for a, b in pairwise(row):
        next_row.append(a + b)
    next_row.append(1)
    return next_row

# list comprehension version of the next_pascal_row function
def next_pascal_row_comp(row):
    return [1] + [a + b for a, b in pairwise(row)] + [1]

row = [1]
print (row)
for i in range(5):
    row = next_pascal_row(row)
    print (row)
```

#### Recursion
Recursion is when a function calls itself to solve a smaller version of the same problem. This continues until it reaches a simple case that can be solved directly, called the base case. The answers are then returned step by step to build the final result.
[Visualization](https://preview.redd.it/i-restored-in-hd-4k-the-original-spider-man-pointing-at-v0-bekphnqftcb41.jpg?width=1080&crop=smart&auto=webp&s=51243691bab29346bb8a94896cf6edba598de3c7)

```python
# recursive version of the next_pascal_row function
def pascal(n):
    if n == 0:
        return [1]

    previous = pascal(n - 1)
    print (previous)

    return [1] + [a + b for a, b in pairwise(previous)] + [1]

pascal(5)
```
