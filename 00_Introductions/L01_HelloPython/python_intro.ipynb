## Again (and again)

A *loop* repeats. This concept is called iteration.

The simplest loop is a `for` loop; `range(5)` means *five times*.

Notice the indentation. The indented lines are the ones that get repeated. Python uses indentation instead of brackets, and it is not optional.

A program is a sequence of instructions

<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/master/content/images/flow_1_sequence.svg" width="240">

Sometimes it is more than just sequential

<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/master/content/images/flow_2_iteration.svg" width="280">

````python
for i in range(3):
    print("knock")

print("who's there?")
````

Three knocks, then one answer.

The knocking is *indented*, so it belongs to the *loop* and happens on every pass. *who's there?* is not indented, so it sits **outside** the loop entirely — it waits until all the knocking has finished, then happens once.

````python
for i in range(5):
    print("*" * 10)
````

Five identical rows. Useful, but not yet interesting — you could have typed five lines.

## One number to rule them all

A shape from a single number Two relationships now: spaces shrink as stars grow.

<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/master/content/images/param_one_number.svg" width="400">

````python
rows = 8

for i in range(rows):
    spaces = " " * (rows - i)
    stars = "*" * (2 * i + 1)
    print(spaces + stars)
````

## Do it (again)!

A **function** gives a name to something you do more than once. You define it once and call it whenever you like.

<img src="https://raw.githubusercontent.com/MAS-dfab/MAS-2627/refs/heads/master/content/images/func_named_box.svg" width="580">

````python
def draw_line(symbol, width):
    print(symbol * width)


draw_line("*", 20)
draw_line("-", 10)
draw_line("*", 20)
````

`symbol` and `width` are the function's parameters — the things you are allowed to vary each time you call it.Deciding *which* things become parameters is a design judgment, not a technical one. It is you deciding what is allowed to change.

````python
def pyramid(rows, symbol):
    for i in range(rows):
        spaces = " " * (rows - i)
        stars = symbol * (2 * i + 1)
        print(spaces + stars)


pyramid(6, "*")
print()
pyramid(10, "#")
````
