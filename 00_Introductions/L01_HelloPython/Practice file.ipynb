```python
#Pascal triangle manual method




def p1(n):
    a = [1]

    for i in range(n): #2
        print(a)
        b = [1] #1

        for i in range(len(a) - 1):
            p = a[i] #1 #2
            q = a[i+1] #2 #1
            r = p + q #3 #3
            b.append(r) #3 #3
        b.append(1)    #1
        a=b



p1(5)

#Pascal triangle definition


```

```python
#Pascal triangle by itertools

from itertools import pairwise

def p3(n):

    x1 = [1]

    for i in range(n):
        print(x1)

        x2 = [1]

        for j,k in pairwise(x1):

            x2.append(j+k)
        x2.append(1)
        x1=x2

p3(5)
```
